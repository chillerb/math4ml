# 2.16 Linear Mappings

> Are the following mappings linear?

> a. Let $a,b \in \mathbb{R}.$
> $$
\begin{aligned}
    \Phi : L^1 ([a,b]) &\mapsto \mathbb{R} \\
    f &\mapsto \Phi(f) = \int\limits_a^b f(x) dx
\end{aligned}$$
> where $L^1(a,b)$ denotes the set of integrable functions on $[a,b]$.

For vector spaces $V, W$, a mapping $\Phi : V \rightarrow W$ is called a *linear mapping* if:

$$
    \forall x, y \in V \forall \lambda, \psi \in \mathbb{R}: \Phi(\lambda x + \psi y) = \lambda \Phi(x) + \psi \Phi(y)
$$

To show this, we can use the linearity of the Integration operation:

$$
    \begin{aligned}
        \forall f, g \in L^1([a,b]) \forall \lambda, \psi \in \mathbb{R}: \Phi(\lambda f + \psi g) &= \int_a^b \lambda f(x) + \psi g(x) dx \\
        &= \int_a^b \lambda f(x) dx + \int_a^b \psi g(x) dx \\
        &= \lambda \int_a^b f(x) dx + \psi \int_a^b g(x) dx \\
        &= \lambda \Phi(f) + \psi \Phi(g) \\
    \end{aligned}
$$

> b. Let $a,b \in \mathbb{R}.$
> $$\begin{aligned}
    \Phi : C^1 &\mapsto C^0 \\
        f &\mapsto \Phi(f) = f'\\
\end{aligned}$$
> where for $k \geq 1$, $C^k$ denotes the set of $k$ times continuously differentiable functions, and $C^0$ denotes the set of continuous functions.

Similar to (a), we can now use the sum and constant rule for differentiation:

$$
    \begin{aligned}
        \forall f, g \in C^1 \forall \lambda, \psi \in \mathbb{R}: \Phi(\lambda f + \psi g) &= \frac{d}{dx} (\lambda f(x) + \psi g(x)) \\
        &= \frac{d}{dx} [\lambda f(x)] + \frac{d}{dx} [\psi g(x)] \\
        &= \lambda \frac{d}{dx} f(x) + \psi \frac{d}{dx} g(x) \\
        &= \lambda \Phi(f) + \psi \Phi(g) \\
    \end{aligned}
$$

> (c) 
> $$\begin{aligned}
    \Phi : \mathbb{R} &\mapsto \mathbb{R} \\
        x &\mapsto \Phi(x) = \cos(x)\\
\end{aligned}$$

Here, we can use the *cosine sum identity*:

$$
    \cos(a + b) = \cos(a)\cos(b) - \sin(a)\sin(b)
$$

and the *Product-to-Sum Identities*:

$$
    \begin{aligned}
        2\cos(a)\cos(b) &= \cos(a - b) + \cos(a+b) \\
        2\sin(a)\sin(b) &= \cos(a - b) - \cos(a+b) \\
    \end{aligned}
$$

$$
    \begin{aligned}
        \forall x, y \in \mathbb{R} \forall \lambda, \psi \in \mathbb{R}: \Phi(\lambda x + \psi y) &= \cos(\lambda x + \psi y)\\
        &= \cos(\lambda x)\cos(\psi y) - \sin(\lambda x)\sin(\psi y)\\
        &= \cos(\lambda x)\cos(\psi y) - \frac{1}{2} [\cos(\lambda x - \psi y) - \cos(\lambda x + \psi y)]\\
        &= \Phi(\lambda x)\Phi(\psi y) - \frac{1}{2} [\Phi(\lambda x - \psi y) - \Phi(\lambda x + \psi y)]\\
        \iff \Phi(\lambda x + \psi y) &= 2 \Phi(\lambda x)\Phi(\psi y) - \Phi(\lambda x - \psi y)
    \end{aligned}
$$

In general, the cosine is not linear, and we can show this by providing a single counter example.

While $\forall n\in \mathbb{Z}: \cos(\theta) = \cos(\theta + n 2 \pi)$, it is not given that $\cos(\lambda \theta) = \lambda \cos(\theta)$.

Let $x, y = \pi / 2$ and $\lambda, \psi = 1$. Then:

$$
    \begin{aligned}
        \Phi(\lambda x + \psi y) &= \cos(\lambda x + \psi y)\\
        &= \cos(\pi)\\
        &= -1\\
        \lambda \Phi(x) + \psi \Phi(y) &= \cos(\pi/2) + \cos(\pi / 2)\\
        &= 0
    \end{aligned}
$$

Therefore, the cosine $\cos$ is not a linear mapping. $\bot$

> (d)
> $$\begin{aligned}
    \Phi : \mathbb{R}^3 &\mapsto \mathbb{R}^2 \\
        \boldsymbol{x} &\mapsto \begin{bmatrix} 1 & 2 & 3\\ 1 & 4 & 3\\\end{bmatrix} \boldsymbol{x}\\
\end{aligned}$$

Let $A = \begin{bmatrix} 1 & 2 & 3\\ 1 & 4 & 3\\\end{bmatrix}$.
Let $z = [\lambda x_1 + \psi y_1, \lambda x_2 + \psi y_2, \lambda x_3 + \psi y_3]^T$.

$$
    \begin{aligned}
        \forall x, y \in \mathbb{R}^3 \forall \lambda, \psi \in \mathbb{R}: \Phi(\lambda x + \psi y) & = A (\lambda x + \psi y) \\
        &= A [\lambda x_1 + \psi y_1, \lambda x_2 + \psi y_2, \lambda x_3 + \psi y_3]^T \\
        &= [1 z_1 + 2 * z_2 + 3 * z_3, 1 * z_1 + 4 * z_2 + 3 * z_3]^T \\
        &= \lambda [x_1 + 2 x_2 + 3x_3, 1 x_1 + 4 x_2 + 3 x_3]^T + \psi [y_1 + 2 y_2 + 3y_3, 1 y_1 + 4 y_2 + 3 y_3]^T \\
        &= \lambda (A x) + \psi (Ay) \\
        &= \lambda \Phi(x) + \psi \Phi(y)
    \end{aligned}
$$

We can show that multiplication for any compatibly shaped matrix $A$ is a linear mapping.
