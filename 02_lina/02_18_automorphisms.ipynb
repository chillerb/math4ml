# 2.18 Automorphisms

> Let $E$ be a vector space. Let $f$ and $g$ be two automorphisms on $E$ such that $f \circ g= \mathrm{id}_E$, i.e., $f\circ g$ is the identity mapping. Show that $\mathrm{ker}(f) = \mathrm{ker}(g\circ f)$, $\mathrm{Im}(g) = \mathrm{Im}(g \circ f)$ and that $\mathrm{ker}(f) \cap \mathrm{Im}(g) = \{ 0_E\}$.

Since $f$ and $g$ are automorphisms, by definition, they are linear and bi-jective (injective and surjective) on $E$:

- *injective*: $\forall x, y \in E: f(x) = f(y) \implies x = y$
- *surjective*: $\forall y \in E \exists x \in E: f(x) = y$

Since $f$ and $g$ are bijective and $f \circ g = \mathrm{id}_E$, it follows that $f$ and $g$ are inverses of each other, i.e.:

$$
    \begin{aligned}
        f(x) &= g^{-1}(x) \\
        g(x) &= f^{-1}(x) \\
    \end{aligned}
$$

And also:

$$
    f \circ g = g \circ f = \mathrm{id}_E
$$

Since $f$ and $g$ are linear:

$$
    f(0) = g(0) = 0
$$

And since $f$ and $g$ are bijective, we have:

$$
    \mathrm{ker}(f) = \mathrm{ker}(g \circ f) = \{ 0\}
$$

Since $g$ is surjective, we have:

$$
    \mathrm{Im}(g) = E
$$

And

$$
    \mathrm{Im}(g \circ f) = \mathrm{Im}(\mathrm{id}_E) = E
$$

Therefore, 

$$
    \mathrm{Im}(g) = \mathrm{Im}(g \circ f) = E
$$

Finally:

$$
    \mathrm{ker}(f) \cap \mathrm{Im}(g) = \{ 0_E\} \cap E = \{0_E \}
$$
