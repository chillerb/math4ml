## Vector Subspaces

> Consider two subspaes $U_1$ and $U_2$, where $U_1$ is the solution space of the homogeneous equation system $A_1 x = 0$, and $U_2$ is the solution space of the homogeneous equation system $A_2 x = 0$ with
> $$
\begin{aligned}
    A_1 &=
    \begin{bmatrix}
        1 & 0 & 1\\
        1 & -2 & -1\\
        2 & 1 & 3\\
        1 & 0 & 1\\
    \end{bmatrix},
    &
    A_2 &=
    \begin{bmatrix}
        3 & -3 & 0\\
        1 & 2 & 3\\
        7 & -5 & 2\\
        3 & -1 & 2\\
    \end{bmatrix}.
\end{aligned}
$$

> (a) Determine the dimension of $U_1$, $U_2$.

$$
\begin{aligned}
    \begin{bmatrix}
        1 & 0 & 1\\
        1 & -2 & -1\\
        2 & 1 & 3\\
        1 & 0 & 1\\
    \end{bmatrix}
    &\leadsto
    \begin{bmatrix}
        1 & 0 & 1\\
        0 & 1 & 1\\
        0 & 0 & 0\\
        0 & 0 & 0\\
    \end{bmatrix}\\
    S_1 &= \left\{\lambda \begin{bmatrix}1\\ 1\\ -1\\\end{bmatrix} \right\}\\
\end{aligned}
$$

$U_1$ is 1-dimensional.

$$
\begin{aligned}
    \begin{bmatrix}
        3 & -3 & 0\\
        1 & 2 & 3\\
        7 & -5 & 2\\
        3 & -1 & 2\\
    \end{bmatrix}
    &\leadsto
    \begin{bmatrix}
        1 & 0 & 1\\
        0 & 1 & 1\\
        0 & 0 & 0\\
        0 & 0 & 0\\
    \end{bmatrix}\\
    S_2 &= \left\{\lambda \begin{bmatrix}1\\ 1\\ -1\\\end{bmatrix} \right\}\\
\end{aligned}
$$

$U_2$ is 1-dimensional.

> (b) Determine the bases of $U_1$ and $U_2$.

$$
\begin{aligned}
    U_1 &= \mathrm{span}[ \begin{bmatrix}1\\ 1\\ -1\\\end{bmatrix} ]\\
    U_2 &= \mathrm{span}[ \begin{bmatrix}1\\ 1\\ -1\\\end{bmatrix} ]\\
\end{aligned}
$$

> (c) Determine the basis of $U_1 \cap U_2$.

In this case, it's trivial to see the basis of the intersection, since $U_1$ and $U_2$ are identical:

$$
    U_1 \cap U_2 = \mathrm{span}[ \begin{bmatrix}1\\ 1\\ -1\\\end{bmatrix} ]\\
$$

```python
import numpy as np

from math4ml import rref, array_to_math
from IPython.display import Markdown

a1 = np.array([[1, 0, 1], [1, -2, -1], [2, 1, 3], [1, 0, 1]])

Markdown(array_to_math(rref(a1)))
```
